# Rainfall problem



- Design a program that consumes a list of numbers (integers) representing daily rainfall amounts as entered by a user.
- The list may contain the number -999 indicating the end of the data of interest.
- Produce the **average of the non-negative values** in the list up to the first -999 (if it shows up).
- There may be negative numbers other than -999 in the list.

### Lösung aus erster Einheit:
- rainfall1

```python
def rainfall1(liste):

    summe = 0
    laenge = 0

    for i in liste:
        if i ==-999:
            break
        if i>=0:
            summe = summe + i
            laenge = laenge + 1  


    try:
        ergebnis = summe/laenge

    except ZeroDivisionError as msg:
        return "Leere Liste (Fehler: {}).".format(msg)
    else: 
        return ergebnis
```

### Mit Filter:

- rainfall2

```python
def rainfall2(liste):

    liste1=[]
    liste2=[]

    for i in liste:
        if i ==-999:
            break
        liste1.append(i)

    def ispositive (x):
        return x>=0

    liste2=list(filter(ispositive,liste1))

    if len(liste2)!=0: 
        return sum(liste2)/len(liste2)

    else:
        return "Leere Liste"
```

### Mit Lambda-Funktion (+ Filter)
- rainfall3

```python
def rainfall3(liste):

    liste1=[]
    liste2=[]

    for i in liste:
        if i ==-999:
            break
        liste1.append(i)

    liste2=list(filter(lambda x:x>=0,liste1))

    if len(liste2)!=0: 
        return sum(liste2)/len(liste2)

    else:
        return "Leere Liste"
```

### Mit While-Loop (ohne Break):
- rainfall4

```python
def rainfall4(liste):
  liste1=[]
  index=0

  while index < len(liste) and liste[index]!=-999:
    if liste[index]>=0:
      liste1.append(liste[index])
    index += 1

  if len(liste1)!=0: 
      return sum(liste1)/len(liste1)

  else:
      return "Leere Liste"
```

### Mit List Comprehension und Slice:
- rainfall5

```python
def rainfall5(liste):

  if -999 in liste:
    liste1=liste[:liste.index(-999)]
  else:
    liste1=liste

  liste2= [x for x in liste1 if x>=0]

  if len(liste2)!=0: 
      return sum(liste2)/len(liste2)

  else:
      return "Leere Liste"
```

### Mit pop():
- rainfall6

### Erklärung: 
liste kopieren, immer das element auf index 0 poppen (kopierteliste.pop(0), und überprüfen ob !=999 und >=0, dann anhängen...


```python
def rainfall6(liste):
   
  liste1=liste.copy()
  liste2 = []

  while len(liste1) >0:
    x=liste1.pop(0)

    if x == -999:
      break

    if x >=0:
      liste2.append(x)

    

  if len(liste2)!=0: 
      return sum(liste2)/len(liste2)

  else:
      return "Leere Liste"
```
